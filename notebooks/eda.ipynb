The purpose of this notebook is to conduct rudimentary analysis of the raw dataset before making the preprocessing pipeline. Because modeling isnt the main point of the project, the eda wont be too detailed

```python
import pandas as pd
import re
import unicodedata
```

```python
df = pd.read_csv("../data/raw/parkinsons_updrs.csv")
```

```python
df.head()
```

```python

def to_snake_case(name: str) -> str:
    name = unicodedata.normalize("NFKD", name)
    name = "".join(c for c in name if not unicodedata.combining(c))
    return re.sub(r"[^a-z0-9]+", "_", name.lower()).strip("_")

df = df.rename(columns=to_snake_case)
```
